```python
# Display top 10 customer name using bar chart
top10_customer = df.groupby("Name")["quantity"].sum().sort_values(ascending=True).head(10)
plt.figure(figsize=(10,5))
sns.barplot(x=top10_customer.index,y=top10_customer.values)
plt.title("Top 10 customer revenue")
plt.xlabel("Customer_Name")
plt.ylabel("quantity")
plt.show()
```

```python
# Display month wise sales revenue
plt.figure(figsize=(12,6))
plt.plot(df["Month"],df["quantity"],marker="o",linestyle="dashed")
plt.title("Month wise sales revenue")
plt.xlabel("Month")
plt.ylabel("quantity")
plt.show()
```

```python
# Display Category wise sales revenue
category = df.groupby("Food_category")["quantity"].sum()
plt.pie(
    category.values,
    labels= category.index,
    autopct="%1.1f%%",
    startangle=45,
    explode=[0.05]*len(category)
)
plt.title("Category wise sales revenue contribution(%)")
plt.show()
    
```

```python
# Display  sales revenue by region 
Region = df.groupby("Region")["quantity"].sum()
plt.pie(
    Region.values,
    labels=Region.index,
    autopct="%1.1f%%",
    startangle=90,
    explode=[0.05]*len(Region)
)
plt.title("Region wise sales revenue contribution(%)")
plt.show()
    
```

```python
# Display the top5 Dish_Name  by sales revenue
TOP5_Dishname = df.groupby("Dish_Name")["quantity"].sum().sort_values(ascending=False).head(5)
sns.barplot(x=TOP5_Dishname.index,y=TOP5_Dishname.values)
plt.title("TOP 5 Dish_name  sale revenue")
plt.xlabel("Dish_Name")
plt.ylabel("sales")
plt.show()
```

```python
# Display Month wise profit revenue
Profit_revenue = df.groupby("Month")["total_profit"].sum().sort_values(ascending=True)
sns.barplot(x=Profit_revenue.index,y=Profit_revenue.values,palette="viridis",legend=False)
plt.title("Month wise profit revenue")
plt.xlabel("Month")
plt.ylabel("total_profit")
plt.xticks(rotation=45)
plt.show()
```

```python
# Display category wise total_profit revenue
Total_profit = df.groupby("Food_category")["total_profit"].sum()
print(Total_profit)
```

```python
#Display category wise average_profit revenue
Average_profit = df.groupby("Food_category")["total_profit"].mean()
print(Average_profit)
```

```python
#Display category wise min_profit revenue
min_profit= df.groupby("Food_category")["total_profit"].min()
print(min_profit)
```

```python
#Display category wise max_profit revenue
Max_profit=df.groupby("Food_category")["total_profit"].max()
print(Max_profit)
```

```python
# Display quantity wise food category revenue using scatter  chart
sns.scatterplot(
   x=df["Food_category"],
   y=df["quantity"],
    hue=df["Food_category"]
)
plt.title("quantity wise Food_category revenue")
plt.show()
```
